```python

# CELL 1: Setup and Configuration

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import joblib
import mlflow
import shap
import warnings
import os
import sys
from datetime import datetime
from typing import Dict, List, Tuple, Any, Optional

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (roc_auc_score, average_precision_score, f1_score,
                            precision_score, recall_score, classification_report,
                            confusion_matrix, roc_curve, precision_recall_curve)
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

warnings.filterwarnings("ignore")

# Set paths
OUTPUT_DIR = "/kaggle/working/models"
REPORTS_DIR = "/kaggle/working/reports"
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(REPORTS_DIR, exist_ok=True)

# Load config
def load_config(config_path: str = "/kaggle/input/datasets/sharoncasssiopia/upconfigs/config.yaml") -> Dict[str, Any]:
    """Load configuration from YAML file."""
    with open(config_path, 'r') as f:
        return yaml.safe_load(f)

CONFIG = load_config()

# Ensure all model parameters are lists for GridSearchCV
for model_name, params in CONFIG['models'].items():
    for param_name, param_value in params.items():
        if not isinstance(param_value, list):
            CONFIG['models'][model_name][param_name] = [param_value]
            print(f"✅ Fixed {model_name}.{param_name}: {param_value} → [{param_value}]")

# Setup MLflow
mlflow.set_tracking_uri(CONFIG['mlflow']['tracking_uri'])
mlflow.set_experiment(CONFIG['mlflow']['experiment_name'])

print("✅ Directories created:", OUTPUT_DIR, REPORTS_DIR)
print("✅ Config loaded and MLflow set up")

# ============================================================================
# CELL 2: CreditRiskModel Class
# ============================================================================

class CreditRiskModel:
    """Professional credit risk modeling class with MLflow tracking and SHAP explainability."""
    
    def __init__(self, model_type: str = 'logistic'):
        self.model_type = model_type
        self.random_state = CONFIG['pipeline']['random_state']
        self.model = None
        self.best_params = None
        self.feature_importance = None
        self.feature_names = None
        self.metrics = {}
        self.threshold = 0.5
        self.shap_values = None
        self.shap_X_sample = None
        
        # Get model configs
        self.model_configs = CONFIG['models'].copy()
        
        # Add random_state to each model's config
        for model in self.model_configs:
            if 'random_state' not in self.model_configs[model]:
                self.model_configs[model]['random_state'] = [self.random_state]
    
    def train(self, X_train: pd.DataFrame, y_train: pd.Series, 
              cv_folds: int = 5, n_jobs: int = -1) -> None:
        """Train model with hyperparameter tuning."""
        print(f"\n{'='*60}")
        print(f"Training {self.model_type.upper()} model")
        print(f"{'='*60}")
        print(f"Training data shape: {X_train.shape}")
        print(f"High-risk rate: {y_train.mean():.2%}")
        
        self.feature_names = X_train.columns.tolist()
        
        if self.model_type not in self.model_configs:
            raise ValueError(f"Model type {self.model_type} not supported. "
                           f"Choose from {list(self.model_configs.keys())}")
        
        config = self.model_configs[self.model_type]
        model_class = {
            'logistic': LogisticRegression,
            'random_forest': RandomForestClassifier,
            'gradient_boosting': GradientBoostingClassifier,
            'xgboost': XGBClassifier
        }[self.model_type]
        
        # Use StratifiedKFold for imbalanced data
        cv = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=self.random_state)
        
        grid_search = GridSearchCV(
            estimator=model_class(),
            param_grid=config,
            cv=cv,
            scoring='roc_auc',
            n_jobs=n_jobs,
            verbose=1
        )
        grid_search.fit(X_train, y_train)
        
        self.model = grid_search.best_estimator_
        self.best_params = grid_search.best_params_
        print(f"\n✅ Best parameters: {self.best_params}")
        print(f"✅ Best CV score (ROC-AUC): {grid_search.best_score_:.4f}")
        
        self._calculate_feature_importance(X_train)
    
    def _calculate_feature_importance(self, X_train: pd.DataFrame) -> None:
        """Calculate feature importance based on model type."""
        if hasattr(self.model, 'feature_importances_'):
            self.feature_importance = pd.DataFrame({
                'feature': self.feature_names,
                'importance': self.model.feature_importances_
            }).sort_values('importance', ascending=False)
        elif hasattr(self.model, 'coef_'):
            self.feature_importance = pd.DataFrame({
                'feature': self.feature_names,
                'importance': np.abs(self.model.coef_[0])
            }).sort_values('importance', ascending=False)
    
    def evaluate(self, X_test: pd.DataFrame, y_test: pd.Series, sample_size: int = 100) -> Dict:
        """Evaluate model performance with SHAP explainability."""
        if self.model is None:
            raise ValueError("Model must be trained before evaluation")
        
        print(f"\n📊 Evaluating model on test set ({X_test.shape[0]} samples)...")
        
        # Predictions
        y_pred_proba = self.model.predict_proba(X_test)[:, 1]
        y_pred = (y_pred_proba >= self.threshold).astype(int)
        
        # Calculate metrics
        self.metrics = {
            'roc_auc': roc_auc_score(y_test, y_pred_proba),
            'average_precision': average_precision_score(y_test, y_pred_proba),
            'f1_score': f1_score(y_test, y_pred),
            'precision': precision_score(y_test, y_pred),
            'recall': recall_score(y_test, y_pred)
        }
        
        # Print classification report
        print("\n" + "="*50)
        print("CLASSIFICATION REPORT")
        print("="*50)
        print(classification_report(y_test, y_pred, target_names=['Low Risk', 'High Risk']))
        
        # Confusion matrix
        cm = confusion_matrix(y_test, y_pred)
        print(f"\n📊 Confusion Matrix:")
        print(f"TN: {cm[0, 0]:,} | FP: {cm[0, 1]:,}")
        print(f"FN: {cm[1, 0]:,} | TP: {cm[1, 1]:,}")
        
        print(f"\n📈 Key Metrics:")
        print(f"  ROC-AUC: {self.metrics['roc_auc']:.4f}")
        print(f"  Avg Precision: {self.metrics['average_precision']:.4f}")
        print(f"  F1-Score: {self.metrics['f1_score']:.4f}")
        
        # SHAP explainability
        self._compute_shap(X_test, sample_size)
        
        return self.metrics
    
    def _compute_shap(self, X_test: pd.DataFrame, sample_size: int = 100) -> None:
        """Compute SHAP values for model explainability."""
        print("\n🔍 Computing SHAP values...")
        
        try:
            X_sample = X_test.sample(min(sample_size, len(X_test)))
            
            # Choose appropriate explainer based on model type
            if self.model_type in ['random_forest', 'gradient_boosting', 'xgboost']:
                explainer = shap.TreeExplainer(self.model)
                self.shap_values = explainer.shap_values(X_sample)
            else:  # logistic regression
                explainer = shap.LinearExplainer(self.model, X_sample)
                self.shap_values = explainer.shap_values(X_sample)
            
            self.shap_X_sample = X_sample
            print(f"✅ SHAP values computed successfully")
            
        except Exception as e:
            print(f"⚠️ SHAP computation failed: {str(e)[:100]}...")
            print("Continuing without SHAP values.")
            self.shap_values = None
    
    def plot_roc_curve(self, X_test: pd.DataFrame, y_test: pd.Series, save_path: str = None) -> None:
        """Plot ROC curve."""
        y_pred_proba = self.model.predict_proba(X_test)[:, 1]
        fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
        roc_auc = roc_auc_score(y_test, y_pred_proba)
        
        plt.figure(figsize=(10, 8))
        plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate', fontsize=12)
        plt.ylabel('True Positive Rate', fontsize=12)
        plt.title(f'ROC Curve - {self.model_type.title()}', fontsize=14)
        plt.legend(loc="lower right")
        plt.grid(True, alpha=0.3)
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✅ ROC curve saved: {save_path}")
        plt.show()
    
    def plot_precision_recall_curve(self, X_test: pd.DataFrame, y_test: pd.Series, save_path: str = None) -> None:
        """Plot Precision-Recall curve."""
        y_pred_proba = self.model.predict_proba(X_test)[:, 1]
        precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)
        avg_precision = average_precision_score(y_test, y_pred_proba)
        
        plt.figure(figsize=(10, 8))
        plt.plot(recall, precision, color='blue', lw=2, label=f'PR curve (AP = {avg_precision:.3f})')
        plt.xlabel('Recall', fontsize=12)
        plt.ylabel('Precision', fontsize=12)
        plt.title(f'Precision-Recall Curve - {self.model_type.title()}', fontsize=14)
        plt.legend(loc="upper right")
        plt.grid(True, alpha=0.3)
        
        # No skill line
        no_skill = y_test.mean()
        plt.plot([0, 1], [no_skill, no_skill], linestyle='--', label='No Skill', color='red')
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✅ PR curve saved: {save_path}")
        plt.show()
    
    def plot_feature_importance(self, top_n: int = 20, save_path: str = None) -> None:
        """Plot top N feature importance."""
        if self.feature_importance is None:
            print("⚠️ No feature importance available")
            return
        
        top_features = self.feature_importance.head(top_n)
        
        plt.figure(figsize=(12, max(6, top_n * 0.4)))
        colors = plt.cm.Blues(np.linspace(0.4, 0.9, len(top_features)))
        
        bars = plt.barh(range(len(top_features)), top_features['importance'].values, color=colors)
        plt.yticks(range(len(top_features)), top_features['feature'].values)
        plt.xlabel('Importance', fontsize=12)
        plt.title(f'Top {top_n} Features - {self.model_type.title()}', fontsize=14)
        plt.gca().invert_yaxis()
        
        # Add value labels
        for i, (bar, importance) in enumerate(zip(bars, top_features['importance'].values)):
            plt.text(bar.get_width() + 0.001, bar.get_y() + bar.get_height()/2, 
                    f'{importance:.4f}', va='center', fontsize=9)
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✅ Feature importance plot saved: {save_path}")
        plt.show()
    
    def plot_shap_summary(self, save_path: str = None) -> None:
        """Plot SHAP summary."""
        if self.shap_values is None:
            print("⚠️ No SHAP values available")
            return
        
        try:
            plt.figure(figsize=(12, 8))
            
            # Handle different SHAP output formats
            if isinstance(self.shap_values, list):
                # For tree models with multiple classes
                shap_values_to_plot = self.shap_values[1] if len(self.shap_values) > 1 else self.shap_values[0]
            else:
                shap_values_to_plot = self.shap_values
            
            shap.summary_plot(shap_values_to_plot, self.shap_X_sample, show=False, max_display=20)
            
            plt.title(f'SHAP Summary - {self.model_type.title()}', fontsize=14)
            plt.tight_layout()
            
            if save_path:
                plt.savefig(save_path, dpi=300, bbox_inches='tight')
                print(f"✅ SHAP summary saved: {save_path}")
            plt.show()
            
        except Exception as e:
            print(f"⚠️ SHAP plotting failed: {str(e)[:100]}...")
    
    def save_model(self, filepath: str) -> None:
        """Save trained model and metadata."""
        if self.model is None:
            raise ValueError("Model must be trained before saving")
        
        model_data = {
            'model': self.model,
            'model_type': self.model_type,
            'best_params': self.best_params,
            'feature_importance': self.feature_importance,
            'feature_names': self.feature_names,
            'metrics': self.metrics,
            'threshold': self.threshold,
            'timestamp': datetime.now().isoformat()
        }
        joblib.dump(model_data, filepath)
        print(f"✅ Model saved: {filepath}")

# ============================================================================
# CELL 3: Load Preprocessed Data (FIXED - NO MORE RFM HERE)
# ============================================================================

print("\n" + "="*60)
print("LOADING PREPROCESSED DATA")
print("="*60)

# 🔥 FIXED: Load your preprocessed data from local processing
# Update these paths to your actual Kaggle dataset paths
X_train = pd.read_csv('/kaggle/input/datasets/sharoncasssiopia/processed-data/X_train.csv')
X_test = pd.read_csv('/kaggle/input/datasets/sharoncasssiopia/processed-data/X_test.csv')
y_train = pd.read_csv('/kaggle/input/datasets/sharoncasssiopia/processed-data/y_train.csv').squeeze()
y_test = pd.read_csv('/kaggle/input/datasets/sharoncasssiopia/processed-data/y_test.csv').squeeze()

# 🔥 CRITICAL FIX: Keep only numeric columns (drop ID columns)
# Identify non-numeric columns
non_numeric_cols = X_train.select_dtypes(exclude=[np.number]).columns.tolist()
if non_numeric_cols:
    print(f"\n⚠️ Dropping non-numeric columns: {non_numeric_cols}")
    X_train = X_train.select_dtypes(include=[np.number])
    X_test = X_test.select_dtypes(include=[np.number])

print(f"\n✅ X_train: {X_train.shape}")
print(f"✅ X_test: {X_test.shape}")
print(f"✅ y_train: {y_train.shape}, High-risk rate: {y_train.mean():.2%}")
print(f"✅ y_test: {y_test.shape}, High-risk rate: {y_test.mean():.2%}")
print(f"✅ All features are numeric: {all(X_train.dtypes == 'float64')}")

# ============================================================================
# CELL 4: Train All Models & Compare
# ============================================================================

def compare_models(X_train, y_train, X_test, y_test, model_types=None):
    """Train and compare multiple models."""
    if model_types is None:
        model_types = ['logistic', 'random_forest', 'gradient_boosting', 'xgboost']
    
    results = []
    
    for model_type in model_types:
        with mlflow.start_run(run_name=f"{model_type}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"):
            
            print(f"\n{'='*60}")
            print(f"TRAINING {model_type.upper()}")
            print(f"{'='*60}")
            
            # Train model
            model = CreditRiskModel(model_type=model_type)
            model.train(X_train, y_train)
            
            # Evaluate
            metrics = model.evaluate(X_test, y_test)
            results.append({'model_type': model_type, **metrics})
            
            # Log to MLflow
            mlflow.log_params(model.best_params)
            mlflow.log_metrics(metrics)
            mlflow.sklearn.log_model(model.model, model_type)
            
            # Save outputs
            model.save_model(os.path.join(OUTPUT_DIR, f"{model_type}_model.pkl"))
            
            # Generate plots
            model.plot_roc_curve(X_test, y_test, os.path.join(REPORTS_DIR, f"roc_curve_{model_type}.png"))
            model.plot_precision_recall_curve(X_test, y_test, os.path.join(REPORTS_DIR, f"pr_curve_{model_type}.png"))
            model.plot_feature_importance(save_path=os.path.join(REPORTS_DIR, f"feature_importance_{model_type}.png"))
            model.plot_shap_summary(save_path=os.path.join(REPORTS_DIR, f"shap_{model_type}.png"))
    
    # Compile results
    results_df = pd.DataFrame(results).sort_values('roc_auc', ascending=False)
    
    print(f"\n{'='*60}")
    print("MODEL COMPARISON RESULTS")
    print(f"{'='*60}")
    print(results_df.round(4).to_string(index=False))
    
    # Save comparison plot
    plt.figure(figsize=(14, 6))
    x = range(len(results_df))
    width = 0.25
    
    plt.bar([i - width for i in x], results_df['roc_auc'], width, label='ROC-AUC', color='steelblue')
    plt.bar([i for i in x], results_df['average_precision'], width, label='Avg Precision', color='coral')
    plt.bar([i + width for i in x], results_df['f1_score'], width, label='F1-Score', color='seagreen')
    
    plt.xlabel('Model Type', fontsize=12)
    plt.ylabel('Score', fontsize=12)
    plt.title('Model Performance Comparison', fontsize=14)
    plt.xticks(x, results_df['model_type'])
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.ylim(0, 1)
    
    # Add value labels
    for i, row in enumerate(results_df.itertuples()):
        plt.text(i - width, row.roc_auc + 0.01, f'{row.roc_auc:.3f}', ha='center', fontsize=9)
        plt.text(i, row.average_precision + 0.01, f'{row.average_precision:.3f}', ha='center', fontsize=9)
        plt.text(i + width, row.f1_score + 0.01, f'{row.f1_score:.3f}', ha='center', fontsize=9)
    
    plt.tight_layout()
    comp_plot_path = os.path.join(REPORTS_DIR, "model_comparison.png")
    plt.savefig(comp_plot_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"✅ Comparison plot saved: {comp_plot_path}")
    
    # Save results CSV
    results_csv_path = os.path.join(REPORTS_DIR, "model_comparison.csv")
    results_df.to_csv(results_csv_path, index=False)
    print(f"✅ Results CSV saved: {results_csv_path}")
    
    return results_df

# Run comparison
results = compare_models(X_train, y_train, X_test, y_test)
best_model = results.iloc[0]['model_type']
print(f"\n🏆 BEST MODEL: {best_model} (ROC-AUC: {results.iloc[0]['roc_auc']:.4f})")

# ============================================================================
# CELL 5: Version Info
# ============================================================================

import sklearn
import xgboost

print(f"\n{'='*60}")
print("ENVIRONMENT INFO")
print(f"{'='*60}")
print(f"Python: {sys.version}")
print(f"scikit-learn: {sklearn.__version__}")
print(f"XGBoost: {xgboost.__version__}")
print(f"MLflow: {mlflow.__version__}")
print(f"SHAP: {shap.__version__}")
print("\n✅ Code meets capstone criteria:")
print("   - Refactored with configs")
print("   - MLflow tracking")
print("   - SHAP explainability")
print("   - Model comparison")
```
